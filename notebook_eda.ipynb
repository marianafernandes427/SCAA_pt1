## Descrição das Variáveis

* **oil_particle_log_count:** Quantidade de partículas de desgaste presentes no lubrificante, numa escala logarítmica.
* **oil_particle_log_iso:** Nível de contaminação do óleo por partículas, ** descobrir a escala**
* **cycle_index:** Número de 0 a 4 do ciclo de funcionamento da máquina
* **post_shutdown_log_delta:** Variação em escala logarítmica após a paragem da máquina, descobrir qual a medida
* **recent_alarm_event_codes:** Históricos de códigos de alarme recentes
* **sensor_channel_constant:** Valor constante de configuração associado a um canal de aquisição de um sensor
* **rpm:** rotações por minuto
* **lubricant_dieletric_constant:** Constante dielética do lubrificante que representa uma propriedade elétrica do óleo que pode ajudar a identificar alterações na composição, contaminação e degradação
* **bearing_clearance_effieciency:** Indicador de eficiência da folga do rolamento, descobri a métrica
* **vibration_radial_rms_mm_s:** Vibração radial RMS em mm/s
* **vibration_axial_rms_mm_s:** Vibração axial RMS em mm/s
* **exhaust_temp_deviation_c:** Desvio da temperatura dos gases do escape
* **temperature_sensor_2_c:** Temperatura do sensor 2
* **machine_serial_id:** Idenficador da série da máquina
* **sensor_channel_dead_99:** Métrica de triagem após um incidente
* **unplanned_outage_next_30d:** Paragem não planeada nos próximos 30 dias
* **75 colunas acerca da telemetria:** 75 bandas ou intervalos de um espectro de sinal de radiofrequência (RF) recolhido pela telemetria da máquina.

```python
import pandas as pd
import numpy as np

# Ajuste dos limites até ao valor máximo de 15 (removendo infinito)
bins = [-0.01, 0.001] + list(range(1, 16))

# Ajuste das etiquetas
labels = ['Exatamente 0']
for i in range(1, 15):
    labels.append(f'({i-1} a {i}]')
labels.append('(14 a 15]')

df_train['post_shutdown_interval'] = pd.cut(df_train['post_shutdown_log_delta'], bins=bins, labels=labels)

# Calcular a média do target por estes novos intervalos detalhados
analise_intervalos = df_train.groupby('post_shutdown_interval', observed=False)['unplanned_outage_next_30d'].agg(['count', 'mean']).reset_index()
analise_intervalos.columns = ['Intervalo (post_shutdown_log_delta)', 'Quantidade de Registos', 'Taxa de Avarias (Média)']

display(analise_intervalos)
```

 dielectric constant decreases as temperature and volume density change, meaning online sensors require a strict temperature-compensation learning phase (typically between 20°C and 80°C) to isolate true degradation from thermal shifts

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Criar intervalos (bins) para a variável 'lubricant_dielectric_constant'
# Analisando os limites: mínimo próximo de 2.0 e máximo próximo de 4.0
bins_dielectric = np.arange(2.0, 4.5, 0.25)
labels_dielectric = [f'({bins_dielectric[i]:.2f} a {bins_dielectric[i+1]:.2f}]' for i in range(len(bins_dielectric)-1)]

# Criar a nova coluna discretizada
df_train['lubricant_dielectric_interval'] = pd.cut(
    df_train['lubricant_dielectric_constant'],
    bins=bins_dielectric,
    labels=labels_dielectric
)

# Exibir a distribuição por intervalos
print("=== DISTRIBUIÇÃO DA CONSTANTE DIELÉTRICA POR INTERVALOS ===")
contagem_intervalos = df_train.groupby('lubricant_dielectric_interval', observed=False)['unplanned_outage_next_30d'].agg(['count', 'mean']).reset_index()
contagem_intervalos.columns = ['Intervalo', 'Quantidade de Registos', 'Taxa de Avarias (Média)']
display(contagem_intervalos)

# 2. Criar um gráfico de dispersão (Scatter Plot)
# Vamos cruzar com a temperatura (temperature_sensor_2_c) para analisar o comportamento térmico e degradação
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df_train,
    x='temperature_sensor_2_c',
    y='lubricant_dielectric_constant',
    hue='unplanned_outage_next_30d',
    palette={0: '#1f77b4', 1: '#d62728'},
    alpha=0.6
)

plt.title('Gráfico de Dispersão: Constante Dielétrica vs Temperatura', fontsize=13, fontweight='bold')
plt.xlabel('Temperatura (Sensor 2) [°C]')
plt.ylabel('Constante Dielétrica do Lubrificante')
plt.legend(title='Avaria (Próximos 30 dias)', labels=['Sem Falha (0)', 'Falha (1)'])
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
import pandas as pd
import numpy as np
import scipy.stats as stats

# 1. Correlação de Pearson apenas para 'sensor_channel_dead_99' com o Target
pearson_corr = df_train['sensor_channel_dead_99'].corr(df_train['unplanned_outage_next_30d'])
print("=== CORRELAÇÃO DE PEARSON (Linear) ===")
print(f"sensor_channel_dead_99 vs unplanned_outage_next_30d: {pearson_corr:.6f}\n")

# 2. Função para o V de Cramér
def cramers_v(x, y):
    confusion_matrix = pd.crosstab(x, y)
    chi2 = stats.chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum().sum()
    phi2 = chi2 / n
    r, k = confusion_matrix.shape
    phi2corr = max(0, phi2 - ((k-1)*(r-1))/(n-1))
    rcorr = r - ((r-1)**2)/(n-1)
    kcorr = k - ((k-1)**2)/(n-1)
    return np.sqrt(phi2corr / min((kcorr-1), (rcorr-1)))

# Calcular V de Cramér apenas para os intervalos de 'sensor_channel_dead_99'
if 'sensor_dead_99_interval' in df_train.columns:
    df_clean = df_train[['sensor_dead_99_interval', 'unplanned_outage_next_30d']].dropna()
    v_cramer = cramers_v(df_clean['sensor_dead_99_interval'], df_clean['unplanned_outage_next_30d'])
    print("=== V DE CRAMÉR (Associação Não-Linear / Por Categorias) ===")
    print(f"sensor_dead_99_interval vs unplanned_outage_next_30d: {v_cramer:.6f}")
else:
    print("A coluna discretizada 'sensor_dead_99_interval' não foi encontrada no DataFrame.")
```

```python
import pandas as pd
import numpy as np
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant

# 1. Selecionar as principais variáveis operacionais numéricas
vars_operacionais = [
    'oil_particle_count_log',
    'oil_particle_count_iso',
    'rpm',
    'lubricant_dielectric_constant',
    'bearing_clearance_efficiency',
    'vibration_radial_rms_mm_s',
    'vibration_axial_rms_mm_s',
    'exhaust_temp_deviation_c',
    'temperature_sensor_2_c',
    'post_shutdown_log_delta'
]

# 2. Criar uma cópia dos dados removendo linhas com nulos apenas nestas colunas
df_vif = df_train[vars_operacionais].dropna().copy()

# 3. Adicionar uma constante para o cálculo correto do VIF
X_vif = add_constant(df_vif)

# 4. Calcular o VIF para cada variável
vif_data = pd.DataFrame()
vif_data["Variável"] = X_vif.columns
vif_data["VIF"] = [variance_inflation_factor(X_vif.values, i) for i in range(X_vif.shape[1])]

# Excluir a constante do resultado final para melhor leitura
vif_data = vif_data[vif_data["Variável"] != "const"].sort_values(by="VIF", ascending=False)

print("=== VARIANCE INFLATION FACTOR (VIF) ===")
display(vif_data)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_selection import mutual_info_classif

# 1. Isolar as colunas de telemetria
cols_rf = [col for col in df_train.columns if 'rf_spectrum_telemetry' in col]

# 2. Criar novas características agregadas a partir do espectro de sinal
df_features = pd.DataFrame(index=df_train.index)

# Média e Desvio Padrão por espectro (por linha)
df_features['rf_mean'] = df_train[cols_rf].mean(axis=1)
df_features['rf_std'] = df_train[cols_rf].std(axis=1)

# Energia total (soma dos quadrados dos sinais - muito comum em análise de vibração/sinal)
df_features['rf_energy'] = df_train[cols_rf].apply(lambda x: np.sum(x**2), axis=1)

# Valor Máximo e Mínimo alcançado no espectro
df_features['rf_max'] = df_train[cols_rf].max(axis=1)
df_features['rf_min'] = df_train[cols_rf].min(axis=1)
df_features['rf_amplitude'] = df_features['rf_max'] - df_features['rf_min']

# Adicionar o Alvo para análise
df_features['unplanned_outage_next_30d'] = df_train['unplanned_outage_next_30d']

# 3. Calcular a correlação linear das novas características
correlacoes_novas = df_features.corr()['unplanned_outage_next_30d'].drop('unplanned_outage_next_30d')
print("=== CORRELAÇÃO LINEAR DAS NOVAS CARACTERÍSTICAS COM O ALVO ===")
print(correlacoes_novas)

# 4. Calcular a Informação Mútua (não-linear)
X_novas = df_features.drop(columns=['unplanned_outage_next_30d'])
y_novas = df_features['unplanned_outage_next_30d']
mi_novas = mutual_info_classif(X_novas, y_novas, random_state=42)

df_mi_novas = pd.DataFrame({
    'Característica': X_novas.columns,
    'Mutual Information': mi_novas
}).sort_values(by='Mutual Information', ascending=False)

print("\n=== INFORMAÇÃO MÚTUA DAS NOVAS CARACTERÍSTICAS ===")
display(df_mi_novas)

# 5. Visualizar a distribuição da Energia Total em relação a falhas
plt.figure(figsize=(10, 5))
sns.boxplot(data=df_features, x='unplanned_outage_next_30d', y='rf_energy', palette='Set2')
plt.title('Distribuição da Energia do Espectro de Telemetria por Estado de Falha')
plt.xlabel('Paragem Não Planeada nos próximos 30 dias (0 = Não, 1 = Sim)')
plt.ylabel('Energia Total do Espectro (Soma dos Quadrados)')
plt.yscale('log') # Escala logarítmica caso haja outliers gigantes de amplitude
plt.show()
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Selecionar apenas colunas numéricas (excluindo o alvo)
numerical_df = df_train.select_dtypes(include=[np.number]).drop(columns=['unplanned_outage_next_30d'], errors='ignore')

# 2. Calcular a matriz de correlação
corr_matrix = numerical_df.corr()

# 3. Identificar pares com alta correlação (limiar de 0.85)
limiar = 0.85
alta_corr = []

# Percorrer a matriz para encontrar os pares redundantes
for i in range(len(corr_matrix.columns)):
    for j in range(i):
        val = corr_matrix.iloc[i, j]
        if abs(val) >= limiar:
            alta_corr.append({
                'Variavel 1': corr_matrix.columns[i],
                'Variavel 2': corr_matrix.columns[j],
                'Correlacao': val
            })

df_alta_corr = pd.DataFrame(alta_corr)

# 4. Mostrar os pares altamente correlacionados
if not df_alta_corr.empty:
    print(f"Encontrados {len(df_alta_corr)} pares de colunas com correlação absoluta >= {limiar}:\n")
    display(df_alta_corr.sort_values(by='Correlacao', ascending=False))

    # 5. Plotar o Heatmap apenas para as colunas redundantes identificadas
    cols_para_plotar = list(set(df_alta_corr['Variavel 1'].tolist() + df_alta_corr['Variavel 2'].tolist()))

    plt.figure(figsize=(12, 10))
    sns.heatmap(
        corr_matrix.loc[cols_para_plotar, cols_para_plotar],
        annot=True,
        fmt=".2f",
        cmap='coolwarm',
        vmin=-1,
        vmax=1,
        square=True,
        linewidths=0.5
    )
    plt.title(f'Matriz de Correlação das Variáveis Altamente Correlacionadas (>= {limiar})', fontsize=14, fontweight='bold')
    plt.show()
else:
    print(f"Nenhum par de variáveis apresentou correlação absoluta acima de {limiar}.")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler

# 1. Selecionar um conjunto representativo de variáveis para o radar
features_radar = [
    'oil_particle_count_log',
    'vibration_axial_rms_mm_s',
    'post_incident_triage_metric',
    'temperature_sensor_2_c',
    'exhaust_temp_deviation_c',
    'bearing_clearance_efficiency',
    'lubricant_dielectric_constant'
]

# Filtrar e remover linhas com valores nulos nestas colunas para o perfil
df_radar = df_train[features_radar + ['unplanned_outage_next_30d']].dropna()

# 2. Normalizar as variáveis numéricas para a escala [0, 1]
scaler = MinMaxScaler()
df_radar[features_radar] = scaler.fit_transform(df_radar[features_radar])

# 3. Calcular a média de cada grupo
perfis = df_radar.groupby('unplanned_outage_next_30d').mean().reset_index()

# 4. Configurar os ângulos para o gráfico de radar
labels = features_radar
num_vars = len(labels)
angles = np.linspace(0, 2 * np.pi, num_vars, endpoint=False).tolist()

# O gráfico de radar precisa ser fechado, então repetimos o primeiro valor
angles += angles[:1]

fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))

# Cores para cada perfil
colors = {0: '#1f77b4', 1: '#d62728'}
labels_mapping = {0: 'Sem Falha (0)', 1: 'Falha Planeada/Inesperada (1)'}

for idx, row in perfis.iterrows():
    target_val = row['unplanned_outage_next_30d']
    values = row[features_radar].values.tolist()
    values += values[:1]

    # Plot da linha
    ax.plot(angles, values, color=colors[target_val], linewidth=2, label=labels_mapping[target_val])
    # Preenchimento da área
    ax.fill(angles, values, color=colors[target_val], alpha=0.25)

# Configurações do layout do radar
ax.set_theta_offset(np.pi / 2)
ax.set_theta_direction(-1)

# Adicionar os rótulos de cada variável
ax.set_thetagrids(np.degrees(angles[:-1]), labels, fontsize=11)

# Título e Legenda
plt.title('Perfil Médio das Características (Escala Normalizada 0-1)', y=1.1, fontsize=14, fontweight='bold')
plt.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))
plt.show()
```
